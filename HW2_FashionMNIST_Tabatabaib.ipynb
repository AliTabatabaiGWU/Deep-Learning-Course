```python
# Load the data

import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras import layers, models

# 1. Load the dataset
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# 2. Flatten and normalize the pixel data
X_train = X_train.reshape(-1, 784).astype("float32") / 255.0
X_test = X_test.reshape(-1, 784).astype("float32") / 255.0

print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")

# establish my own MLP  architecture 2 hidden layers (128 to 64), activation RELU for hidden layers SOFTMAX for output layers (10 classes_), optimizer Adam, learning rate 0.001. batch size 64, epoch 10  we keep these constants to control a consistent benchmark for the performance comparison

def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"Baseline Test Accuracy: {test_acc:.4f}")

# design and run 6 abration experiments  activation from RELU to sigmoid, optimizer from adam to sgd, change the architecture to single hiden layer of 64, learning rate from 0.001 to 0.01, batch size from 64 to 256,  drop out to 0.3


# Case 1 Run the experiment with optimizer as SGD

def build_baseline_architecture():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

sgd_model = build_baseline_architecture()

# Compile with SGD optimizer
sgd_model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.001), # Changed only this variable
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print the  model summary
sgd_model.summary()

# Train the c model variable
history_sgd = sgd_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2,
    verbose=1
)

# Evaluate the model variable
test_loss, test_acc = sgd_model.evaluate(X_test, y_test, verbose=0)
print(f"Case 1 SGD Model Test Accuracy: {test_acc:.4f}")

# Case 2 change learning rate to 0.01


def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"Case 2 learning rate 0.01  Test Accuracy: {test_acc:.4f}")

# case 3 use single layer

def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"Case 3_one hidden layer 64 Test Accuracy: {test_acc:.4f}")


# Case 4  Change Relu to Sigmoid

def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='sigmoid'),
        layers.Dense(64, activation='sigmoid'),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"Case 4 sigmoid Test Accuracy: {test_acc:.4f}")

# case 5 change batch size from 64 to 128


def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"case 5 batch size 128 test Accuracy: {test_acc:.4f}")

# Case 6  Change dropour tae from o to 30%


def build_baseline_model():
    model = models.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(10, activation='softmax')
    ])
    return model

# Setup baseline settings

baseline_model = build_baseline_model()
baseline_model.compile(
      optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
      loss='sparse_categorical_crossentropy',
     metrics=['accuracy']
)
#
#dropout_model
baseline_model.compile(
  optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
  loss='sparse_categorical_crossentropy',
  metrics=['accuracy']
)
# Print total parameter count for   table
baseline_model.summary()

# Train with a validation split

history = baseline_model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2, # 20% validation split
    verbose=1
)

# Evaluate on the test set

test_loss, test_acc = baseline_model.evaluate(X_test, y_test, verbose=0)
print(f"Case 6 dropout set to 0.3 Test Accuracy: {test_acc:.4f}")

```
